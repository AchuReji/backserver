Bank Account Manager


```python

account_balance = 500.00

def display_balance():

    print(f"\n--- Balance Enquiry ---")

    print(f"Current Account Balance: ₹{account_balance:,.2f}")
    return account_balance

def deposit(amount):

    global account_balance
    if amount > 0:
        # Perform the deposit
        account_balance += amount
        print(f"\n--- Deposit Successful ---")
        print(f"Deposited: ₹{amount:,.2f}")
        display_balance()
    else:
        print("\nInvalid deposit amount. Amount must be positive.")



```

```python
def withdraw(amount):

    global account_balance
    print(f"\n--- Withdrawal Attempt ---")
    print(f"Requested Withdrawal: ₹{amount:,.2f}")

    if amount <= 0:
        print("Invalid withdrawal amount. Amount must be positive.")
        return


    if amount <= account_balance:

        account_balance -= amount
        print("Withdrawal successful.")
        display_balance()
    else:
        print("Transaction failed: Insufficient Balance!")
        display_balance()
```

```python
print("--- Bank Account Manager Simulation Started ---")


display_balance()


deposit(150.75)  # Successful deposit

withdraw(50.00)  # Successful withdrawal

withdraw(1000.00) # Failed withdrawal

display_balance() # Final balance check
```

Weather Data Analyzer

```python

region_name = "Kochi" # string
temperatures = [32.5, 30.1, 31.8, 29.9]
humidities = [75, 80, 72, 85, 78]

def calculate_averages(temps, hums):

    # Calculate average temperature
    avg_temp = sum(temps) / len(temps) if temps else 0

    # Calculate average humidity
    avg_humidity = sum(hums) / len(hums) if hums else 0

    print(f"\n--- Calculated Averages for {region_name} ---")
    print(f"Average Temperature: {avg_temp:.2f} °C")
    print(f"Average Humidity: {avg_humidity:.2f}%")

    return avg_temp, avg_humidity

```

```python
def determine_weather_conditions(temp, hum):

    condition = ""

    # Analyze Temperature
    if temp >= 32.0:
        condition += "Hot "
    elif temp >= 28.0:
        condition += "Warm "
    else:
        condition += "Pleasant "

    # Analyze Humidity (using a combined check for simplicity)
    if hum >= 80:
        condition += " and Very Humid 💧"
    elif hum >= 70:
        condition += " and Moderately Humid"
    else:
        condition += " and Low Humidity"

    print(f"The determined condition is: {condition}")
```

```python



# d.  calculate average temperature and humidity
average_temp, average_humidity = calculate_averages(temperatures, humidities)

#   weather conditions
determine_weather_conditions(average_temp, average_humidity)
```
